**Завдання 9**. Натренуйте лінійну регресію з `statsmodels` тільки на ознаках, які виявлись стат. значущими в попередньому завданні. Проаналізуйте показники моделі. Чи значно змінились R2 і Adj. R-squared?

````python
Y = y_train
significant_features_1 = ['Mileage', 'Power', 'Fuel_Type_code', 'Transmission_code','Audi',
                        'BMW', 'Ford', 'Hyundai', 'Mahindra', 'Maruti', 'Mercedes', 'Volkswagen']
X_full = pd.DataFrame(scaled_X_train, columns=X_train.columns, index=X_train.index)

X = X_full[significant_features_1]
X = sm.add_constant(X)
model = sm.OLS(Y,X)
results = model.fit()
results.params

print(results.summary())
````
